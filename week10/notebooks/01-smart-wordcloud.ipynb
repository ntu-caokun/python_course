### 交互式（课堂上实时输入）

```python
# ========== 方式三：交互式（课堂上实时输入） ==========
# 这段代码可以作为课堂演示，让学生现场投票
print("=" * 50)
print("请依次输入你认为与“智能”相关的关键词（输入q结束）：")
print("=" * 50)

user_words = []
while True:
    word = input("> ")
    if word.lower() == 'q':
        break
    if word.strip():
        user_words.append(word.strip())

# 统计并生成词云
if user_words:
    user_freq = Counter(user_words)
    wordcloud3 = WordCloud(
        font_path='/System/Library/Fonts/PingFang.ttc',  # macOS
        # font_path='C:/Windows/Fonts/simhei.ttf',      # Windows
        width=1000,
        height=600,
        background_color='white',
        colormap='coolwarm'
    ).generate_from_frequencies(user_freq)
    
    fig3, ax3 = plt.subplots(figsize=(14, 8))
    ax3.imshow(wordcloud3, interpolation='bilinear')
    ax3.axis('off')
    # ax3.set_title('课堂实时生成的“智能”词云', fontsize=20, fontweight='bold', pad=20)
    plt.tight_layout()
    plt.show()
else:
    print("未输入任何关键词。")
```
