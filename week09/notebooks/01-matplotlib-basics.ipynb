## 两种绘图方式

### pyplot 快速模式（适合简单绘图）

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y)
plt.title("Sine Wave")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.show()
```

### 面向对象模式（推荐，更灵活）

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(x, y)
ax.set_title("Sine Wave")
ax.set_xlabel("x")
ax.set_ylabel("sin(x)")
plt.show()
```

## 常用图表类型

### 折线图

```python
# 绘制多条折线
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

fig, ax = plt.subplots(figsize=(8, 8))
ax.plot(x, y1, label='sin(x)', color='blue', linestyle='-')
ax.plot(x, y2, label='cos(x)', color='red', linestyle='--')
ax.legend()
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Sine and Cosine')
plt.show()
```

### 散点图

```python
# 随机数据散点图
np.random.seed(42)
x = np.random.randn(100)
y = 2*x + np.random.randn(100)

fig, ax = plt.subplots()
ax.scatter(x, y, alpha=0.6, color='steelblue')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_title('Scatter Plot')
plt.show()
```

### 柱状图

```python
# 垂直柱状图
categories = ['A', 'B', 'C', 'D']
values = [23, 45, 56, 78]

fig, ax = plt.subplots()
ax.bar(categories, values, color='skyblue')
ax.set_xlabel('Category')
ax.set_ylabel('Value')
ax.set_title('Bar Chart')
plt.show()

# 水平柱状图
fig, ax = plt.subplots()
ax.barh(categories, values, color='lightcoral')
ax.set_xlabel('Value')
ax.set_ylabel('Category')
ax.set_title('Horizontal Bar Chart')
plt.show()
```

### 直方图

```python
# 正态分布数据直方图
data = np.random.randn(1000)

fig, ax = plt.subplots()
ax.hist(data, bins=30, edgecolor='black', alpha=0.7, color='green')
ax.set_xlabel('Value')
ax.set_ylabel('Frequency')
ax.set_title('Histogram')
plt.show()
```

### 饼图

```python
# 市场份额饼图
sizes = [30, 25, 20, 15, 10]
labels = ['Python', 'Java', 'C++', 'JavaScript', 'Other']
colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99', '#c2c2f0']

fig, ax = plt.subplots()
ax.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90)
ax.set_title('Market Share')
plt.show()
```

### 图例和网格

```python
import matplotlib.pyplot as plt
import numpy as np

# 生成数据
x = np.linspace(0, 10, 100)   # 从 0 到 10 均匀生成 100 个点
y1 = np.sin(x)                 # sin(x)
y2 = np.cos(x)                 # cos(x)

# 绘图
fig, ax = plt.subplots()
ax.plot(x, y1, label='sin(x)')
ax.plot(x, y2, label='cos(x)')
ax.legend(loc='upper right', fontsize=10, framealpha=0.5)
ax.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

## 子图

### 使用 subplots()

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# 生成数据
x = np.linspace(0, 10, 100)   # 从 0 到 10 均匀生成 100 个点
y1 = np.sin(x)                 # sin(x)

# 第一行第一列
axes[0, 0].plot(x, y1)
axes[0, 0].set_title('Sine')

# 第一行第二列
axes[0, 1].scatter(x, y1)
axes[0, 1].set_title('Sine Scatter')

# 第二行第一列
axes[1, 0].hist(y1, bins=20)
axes[1, 0].set_title('Sine Histogram')

# 第二行第二列
axes[1, 1].boxplot(y1)
axes[1, 1].set_title('Sine Boxplot')

plt.tight_layout()
plt.show()
```

### 不规则子图（add_subplot）

```python
import matplotlib.pyplot as plt
import numpy as np

# 生成示例数据
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.exp(-x/5) * np.sin(x)  # 衰减振荡

# 创建不规则子图布局
fig = plt.figure(figsize=(10, 6))

# 位置1：2行2列的第1个（左上）
ax1 = fig.add_subplot(2, 2, 1)
ax1.plot(x, y1, 'b-')
ax1.set_title('sin(x)')
ax1.set_xlabel('x')
ax1.set_ylabel('y')
ax1.grid(True, linestyle='--', alpha=0.3)

# 位置2：2行2列的第2个（右上）
ax2 = fig.add_subplot(2, 2, 2)
ax2.plot(x, y2, 'r-')
ax2.set_title('cos(x)')
ax2.set_xlabel('x')
ax2.set_ylabel('y')
ax2.grid(True, linestyle='--', alpha=0.3)

# 位置3：2行1列的第2个（占整个下半部分）
ax3 = fig.add_subplot(2, 1, 2)
ax3.plot(x, y3, 'g-')
ax3.set_title('Damped Oscillation: exp(-x/5) * sin(x)')
ax3.set_xlabel('x')
ax3.set_ylabel('y')
ax3.grid(True, linestyle='--', alpha=0.3)

# 自动调整子图间距
plt.tight_layout()
plt.show()
```

## 保存图形

```python
fig, ax = plt.subplots()
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
ax.plot(x, y1)
ax.set_title('My Plot')

# 保存为不同格式
fig.savefig('my_plot.png', dpi=300, bbox_inches='tight')    # 可调分辨率，图片格式
fig.savefig('my_plot.pdf')
fig.savefig('my_plot.svg')
```

## 中文显示

```python
# 解决 Matplotlib 中文显示问题
plt.rcParams['font.sans-serif'] = ['Arial Unicode MS', 'SimHei', 'WenQuanYi Micro Hei']
plt.rcParams['axes.unicode_minus'] = False

# 现在可以显示中文了
fig, ax = plt.subplots()
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
ax.plot(x, y1, label='正弦函数')
ax.set_title('正弦波')
ax.set_xlabel('x 轴')
ax.set_ylabel('y 轴')
ax.legend()
plt.show()
```
