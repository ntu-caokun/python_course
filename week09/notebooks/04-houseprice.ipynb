## 数据分析

```python
# 统计摘要打印（环比）
print("\n" + "="*60)
print("【环比统计摘要】")
print("="*60)
print(f"环比>100的城市（上涨）：{(df['环比'] > 100).sum()}个")
print(f"环比=100的城市（持平）：{(df['环比'] == 100).sum()}个")
print(f"环比<100的城市（下跌）：{(df['环比'] < 100).sum()}个")

# 最大值及其城市
max_val = df['环比'].max()
max_cities = df[df['环比'] == max_val]['城市'].tolist()
print(f"环比最大值：{max_val}")
print(f"  对应城市：{max_cities}（共{len(max_cities)}个）")

# 最小值及其城市
min_val = df['环比'].min()
min_cities = df[df['环比'] == min_val]['城市'].tolist()
print(f"环比最小值：{min_val}")
print(f"  对应城市：{min_cities}（共{len(min_cities)}个）")

print(f"环比均值：{df['环比'].mean():.2f}")
print(f"环比中位数：{df['环比'].median():.2f}")

# 统计摘要打印（同比）
print("\n" + "="*60)
print("【同比统计摘要】")
print("="*60)
print(f"同比>100的城市：{(df['同比'] > 100).sum()}个")

# 最大值及其城市
max_val = df['同比'].max()
max_cities = df[df['同比'] == max_val]['城市'].tolist()
print(f"同比最大值：{max_val}")
print(f"  对应城市：{max_cities}（共{len(max_cities)}个）")

# 最小值及其城市
min_val = df['同比'].min()
min_cities = df[df['同比'] == min_val]['城市'].tolist()
print(f"同比最小值：{min_val}")
print(f"  对应城市：{min_cities}（共{len(min_cities)}个）")
```

```python
# 分类
double_up = (df['环比'] > 100) & (df['同比'] > 100)
recovery = (df['环比'] > 100) & (df['同比'] < 100)
double_down = (df['环比'] < 100) & (df['同比'] < 100)
pullback = (df['环比'] < 100) & (df['同比'] > 100)

# 四象限城市列表（供课堂讲解使用）
print("\n" + "="*60)
print("【四象限分类】")
print("="*60)
print(f"双涨（环比>100且同比>100）：{list(df.loc[double_up, '城市'])}")
print(f"短期回暖（环比>100但同比<100）：{list(df.loc[recovery, '城市'])}")
print(f"双跌（环比<100且同比<100）：{len(df.loc[double_down, '城市'])}个城市")
print(f"短期回调（环比<100但同比>100）：{list(df.loc[pullback, '城市'])}")
```

## 环比排序条形图

```python
# 环比排序条形图
fig, ax = plt.subplots(figsize=(14, 20))
df_sorted = df.sort_values('环比')
colors = ['#d62728' if x > 100 else '#1f77b4' if x < 100 else '#2ca02c' for x in df_sorted['环比']]
ax.barh(df_sorted['城市'], df_sorted['环比'] - 100, color=colors)
ax.axvline(x=0, color='black', linewidth=1.5, linestyle='-')
ax.set_xlabel('环比涨跌幅（百分点）', fontsize=14)
ax.set_title('2026年3月70个大中城市新建商品住宅环比涨跌排序', fontsize=16, fontweight='bold')
ax.grid(axis='x', alpha=0.4)
ax.text(0.02, 0.98, '红色=上涨  蓝色=下跌  绿色=持平', transform=ax.transAxes, fontsize=12, 
        bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(20, 10))  # 宽度加大，高度减小（因为城市在X轴）
df_sorted = df.sort_values('环比', ascending=False)
colors = ['#d62728' if x > 100 else '#1f77b4' if x < 100 else '#2ca02c' for x in df_sorted['环比']]

# 改为垂直条形图 bar()
ax.bar(df_sorted['城市'], df_sorted['环比'] - 100, color=colors)

# 参考线（水平线，因为现在是垂直条形图）
ax.axhline(y=0, color='black', linewidth=1.5, linestyle='-')

ax.set_ylabel('环比涨跌幅（百分点）', fontsize=14)
ax.set_xlabel('城市', fontsize=14)
ax.set_title('2026年3月70个大中城市新建商品住宅环比涨跌排序', fontsize=16, fontweight='bold')

# 网格改为Y轴方向
ax.grid(axis='y', alpha=0.4)

# 图例文本
ax.text(0.85, 0.95, '红色=上涨  蓝色=下跌', transform=ax.transAxes, fontsize=18, 
        bbox=dict(boxstyle='round', facecolor='white', alpha=0.4))

plt.xticks(rotation=45, ha='right', fontsize=10)        # 旋转X轴标签，避免重叠

plt.tight_layout()
plt.show()
```

## 环比-同比四象限散点图

```python
# 环比-同比四象限散点图
fig, ax = plt.subplots(figsize=(16, 10))
# 分类
double_up = (df['环比'] > 100) & (df['同比'] > 100)
recovery = (df['环比'] > 100) & (df['同比'] <= 100)
double_down = (df['环比'] < 100) & (df['同比'] < 100)
pullback = (df['环比'] < 100) & (df['同比'] >= 100)
# 参考线
ax.axhline(y=100, color='gray', linestyle='--', linewidth=1.5)
ax.axvline(x=100, color='gray', linestyle='--', linewidth=1.5)
# 绘制
ax.scatter(df.loc[double_up, '环比'], df.loc[double_up, '同比'], color='#d62728', s=120, label='双涨', alpha=0.8)
ax.scatter(df.loc[recovery, '环比'], df.loc[recovery, '同比'], color='#ff7f0e', s=120, label='短期回暖', alpha=0.8)
ax.scatter(df.loc[double_down, '环比'], df.loc[double_down, '同比'], color='#1f77b4', s=120, label='双跌', alpha=0.8)
ax.scatter(df.loc[pullback, '环比'], df.loc[pullback, '同比'], color='#2ca02c', s=120, label='短期回调', alpha=0.8)
# 显示城市
highlight = df['城市'].tolist()
for city in highlight:
    row = df[df['城市'] == city].iloc[0]
    ax.annotate(city, (row['环比'], row['同比']), xytext=(8, -3), textcoords='offset points', fontsize=12, fontweight='bold')
# 坐标、图例、标题等
ax.set_xlabel('环比指数 (上月=100)', fontsize=16)
ax.set_ylabel('同比指数 (上年同月=100)', fontsize=16)
ax.set_title('70城房价四象限图：环比 vs 同比', fontsize=16, fontweight='bold')
ax.legend(loc='upper left', fontsize=16)
ax.grid(alpha=0.3)
ax.set_xlim(99, 101)
ax.set_ylim(92, 105)
ax.tick_params(axis='both', labelsize=14) # 同时设置X轴和Y轴的刻度标签字体大小
# 布局
plt.tight_layout()
plt.show()
```
